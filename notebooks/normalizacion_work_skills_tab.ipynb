```python
import pandas as pd
import numpy as np
```

```python
df_work = pd.read_csv('../data/split_tabs/work.csv')
df_work.head(5)
```

```python
df_skills_candidate = pd.read_csv('../data/split_tabs/skills_candidate.csv')
```

```python
df_work_skills_temp = df_work.merge(df_skills_candidate,'inner',on='candidate_id')
df_work_skills_temp.head(5)
```

```python
df_skills = pd.read_csv('../data/split_tabs/skills.csv')
```

```python
df_work_skills = df_work_skills_temp.merge(df_skills,'inner',on='skill_name')
df_work_skills.head(15)
```

```python
df_work_skills[df_work_skills['skill_name']=='Python']
```

```python
df_work_skills[df_work_skills['candidate_id']==1]
```

```python
df_work_skills = df_work_skills[['work_id', 'skills_id']]
df_work_skills.head(5)
```

```python
df_work_skills[df_work_skills['skills_id']==1]
```

```python
df_work_skills.index.name = 'work_skills_id'
```

```python
df_work_skills.to_csv('../data/split_tabs/work_skills.csv')
df_work_skills.head(5)
```
