```python
# Rutas de los archivos CSV
paths = {
    # 'main_data': '../data/processed/victimas_por_hechos_departamental_20250416.csv'
    'candidate': '../data/split_tabs/candidate.csv',
    'educational_information': '../data/split_tabs/educational_information.csv',
    'recruitment_information': '../data/split_tabs/recruitment_information.csv',
    'skills': '../data/split_tabs/skills.csv',
    'work_skills': '../data/split_tabs/work_skills.csv',
    'work': '../data/split_tabs/work.csv',
      # reporte_hecho y grupo_poblacional misma PK
}

# Diccionario para guardar los DataFrames cargados
dfs = {}

for name, path in paths.items():
    try:
        # Intentar cargar con UTF-8
        df = pd.read_csv(path, encoding='utf-8')
    except UnicodeDecodeError:
        try:
            # Intentar cargar con Latin1 si falla UTF-8
            df = pd.read_csv(path, encoding='latin1')
        except Exception as e:
            print(f"❌ Error cargando '{name}' desde '{path}': {e}")
            df = None
    dfs[name] = df

# Mostrar resumen de la carga
print("\n📦 Resumen de archivos cargados:")
for name, df in dfs.items():
    if df is not None:
        print(f"✅ '{name}' cargado con éxito. Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
    else:
        print(f"⚠️ '{name}' no se cargó correctamente.")
```

```python
dfs['candidate']
```

```python
cursor = conexion.cursor()

df_candidate = dfs['candidate']

# Iterar sobre las filas del DataFrame
for _, row in df_candidate.iterrows():
    try:
        query = """
            INSERT INTO candidate(candidate_id, age, gender, country)
            VALUES (%s, %s,%s, %s)
        """
        values = (int(row['candidate_id']), row['candidate_id'])  # Asegurar el tipo entero
        cursor.execute(query, values)
    except Exception as e:
        print(f"❌ Error insertando el candidate con ID {row['candidate_id']}: {e}")

# Confirmar los cambios
conexion.commit()
print("✅ Inserción completada en candidate.")

cursor.close()
```
