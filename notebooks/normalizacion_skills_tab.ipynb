```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/processed/stackoverflow_full.csv')
```

```python
df_skills = df['HaveWorkedWith']
df_skills.head(5)
```

```python
df_skills = df['HaveWorkedWith'].str.split(';',expand=True).stack()
df_skills.tail(10)
```

```python
df_skills = df_skills.reset_index()
df_skills.columns = ['candidate_id','skill_position', 'skill_name']
df_skills = df_skills[['candidate_id','skill_name']]
df_skills.head(10)
```

```python
df_skills['skill_name'] = df_skills['skill_name'].str.strip()
df_skills.tail(10)
```
