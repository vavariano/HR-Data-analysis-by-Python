```python
# Extraer skills únicas de df_skills
df_skills_catalog = df_skills[['skill_name']].drop_duplicates().reset_index(drop=True)

# Agregar IDs
df_skills_catalog['skills_id'] = range(0, len(df_skills_catalog))

# Reordenar columnas
df_skills_catalog = df_skills_catalog[['skills_id', 'skill_name']]

df_skills_catalog.tail(10)

df_skills_catalog[df_skills_catalog['skill_name'] == 'Python']
```

```python
df_skills_catalog.to_csv('../data/split_tabs/skills.csv',index=False)
```
